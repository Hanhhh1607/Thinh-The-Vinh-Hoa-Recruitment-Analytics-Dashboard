#ROUN2: DATA CLEANING AND STRUCTURING

```python
ops_raw = pd.read_excel(path_file2, sheet_name='OPS')
ops = ops_raw[ops_raw['Họ và tên'].notna()].copy()
print(f"OPS: {len(ops_raw)} dòng thô -> {len(ops)} dòng có dữ liệu thật")
```

```python
print(list(ops.columns))
```

```python
for c in ['Vòng 2 Ngày phỏng vấn SM','Vòng 2 Ngày phỏng vấn SM.1','Vòng 2 Ngày phỏng vấn SM.2']:
    print(c, "-> non-null:", ops[c].notna().sum())
    print(ops[c].dropna().head(2).tolist())
    print()
```

```python
ops['Ngay_PV_Vong2_raw'] = ops['Vòng 2 Ngày phỏng vấn SM.2'].combine_first(
                                ops['Vòng 2 Ngày phỏng vấn SM'])
ops['Gio_PV_Vong2_raw'] = ops['Vòng 2 Thời gian phỏng vấn'].combine_first(
                               ops['Vòng 2 Ngày phỏng vấn SM.1'])
ops['KQ_PV_Vong2_raw'] = ops['Kết quả phỏng vấn vòng 2']
ops['Ngay_Nhan_Viec_raw'] = ops['Ngày nhận việc.1'].combine_first(ops['Ngày nhận việc'])
ops['Vi_Tri_Nhan_Viec_raw'] = ops['Vị trí nhận việc.1'].combine_first(ops['Vị trí nhận việc'])
ops['Pass_Vong2_Flag_raw'] = ops[' ']
ops['Onboard_Note_raw'] = ops['Đến nhận việc không?.1']

print("Ngay_PV_Vong2_raw non-null:", ops['Ngay_PV_Vong2_raw'].notna().sum())
```

```python
def to_date_safe(series):
    s = pd.to_datetime(series, errors='coerce')
    s = s.where((s.dt.year >= 2024) & (s.dt.year <= 2026), pd.NaT)
    return s

date_cols_map = {
    'Ngay_Ung_Tuyen':   'Ngày ứng tuyển',
    'Ngay_So_Van':      'Ngày sơ vấn',
    'Ngay_PV_Vong2':    'Ngay_PV_Vong2_raw',
    'Ngay_Nhan_Viec':   'Ngay_Nhan_Viec_raw',
}
for new_col, raw_col in date_cols_map.items():
    ops[new_col] = to_date_safe(ops[raw_col])
    ops[f'{new_col}_Status'] = np.where(
        ops[raw_col].notna() & ops[new_col].isna(), 'Invalid',
        np.where(ops[new_col].notna(), 'Valid', 'Blank'))
```

```python
for col in date_cols_map:
    print(col, "->", ops[f'{col}_Status'].value_counts().to_dict())
```

```python
def to_time_text(v):
    if pd.isna(v): return np.nan
    if hasattr(v, 'strftime'):
        try: return v.strftime('%H:%M')
        except Exception: pass
    return str(v).strip()

ops['Gio_PV_Vong2'] = ops['Gio_PV_Vong2_raw'].apply(to_time_text)

nam_sinh_str = ops['Năm sinh'].astype(str).str.strip()
year_extracted = nam_sinh_str.str.extract(r'(\d{4})$')[0]
valid_birth_year = year_extracted.notna() & year_extracted.astype(float).between(1960, 2012)
nam_sinh_str = nam_sinh_str.where(valid_birth_year)
ops['Ngay_Sinh'] = pd.to_datetime(nam_sinh_str, format='%d/%m/%Y', errors='coerce')
ops['Tuoi_Ung_Tuyen'] = np.where(
    ops['Ngay_Sinh'].notna() & ops['Ngay_Ung_Tuyen'].notna(),
    (ops['Ngay_Ung_Tuyen'] - ops['Ngay_Sinh']).dt.days // 365, np.nan)
```

```python
print("Tuổi min/max:", ops['Tuoi_Ung_Tuyen'].min(), ops['Tuoi_Ung_Tuyen'].max())
```

```python
import re
def norm_text(v):
    if pd.isna(v): return np.nan
    return str(v).strip()

ops['KQ_So_Van'] = ops['Kết quả sơ vấn'].apply(norm_text)
ops['KQ_PV_Vong2'] = ops['KQ_PV_Vong2_raw'].apply(norm_text)
ops['Pass_Vong2_Flag'] = ops['Pass_Vong2_Flag_raw'].apply(norm_text).map(
    {'x': 'Pass', 'X': 'Pass', 'Reject': 'Reject'})

def clean_phone(s):
    if pd.isna(s): return np.nan
    s = re.sub(r'\D', '', str(s))
    if s.startswith('84') and len(s) == 11: s = '0' + s[2:]
    if not s.startswith('0') and len(s) == 9: s = '0' + s
    if len(s) != 10: return np.nan
    return s

ops['SDT'] = ops['Số điện thoại'].apply(clean_phone)
print("SĐT hợp lệ:", ops['SDT'].notna().sum(), "/", len(ops))
```

##CASE1: DROP-OFF PHỄU TUYỂN DỤNG

```python
ops['Co_SoVan']  = ops['KQ_So_Van'].notna()
ops['Dat_SoVan'] = ops['KQ_So_Van'] == 'Đạt'

ops['Co_PV2']  = ops['KQ_PV_Vong2'].notna() & (ops['KQ_PV_Vong2'] != '-')
ops['Dat_PV2'] = ops['KQ_PV_Vong2'] == 'Đạt'

print("Có sơ vấn:", ops['Co_SoVan'].sum(), "| Đạt sơ vấn:", ops['Dat_SoVan'].sum())
print("Có PV2:", ops['Co_PV2'].sum(), "| Đạt PV2:", ops['Dat_PV2'].sum())
```

=> Loại '-' ở Co_PV2: cột Kết quả phỏng vấn vòng 2 có 1 dòng giá trị là dấu gạch ngang -, đây là dữ liệu rỗng được nhập nhầm dạng text chứ không phải kết quả thật — nếu không loại, dòng đó sẽ tính nhầm là "có PV2" trong mẫu số tỷ lệ.

```python
reason_map = {
    'Ứng viên không đến phỏng vấn': 'No-show PV',
    'ứng viên không đến phỏng vấn': 'No-show PV',
    'Không đến phỏng vấn': 'No-show PV',
    'Chưa phù hợp': 'Không phù hợp (kỹ năng/JD)',
    'Không đạt': 'Không phù hợp (kỹ năng/JD)',
    'Ứng viên dời lịch PV': 'Dời lịch / chưa chốt',
    'Ứng viên dời lịch sang tuần kế': 'Dời lịch / chưa chốt',
    'Xin dời PV': 'Dời lịch / chưa chốt',
    'UV dời PV': 'Dời lịch / chưa chốt',
    'Từ chối PV': 'Ứng viên từ chối',
    'Reject PV': 'Ứng viên từ chối',
    'Cân nhắc': 'Đang cân nhắc',
    'Pending': 'Đang cân nhắc',
    'Chờ phỏng vấn': 'Đang cân nhắc',
    'Khu vực không có HC': 'Không có headcount khu vực',
    'Chuyển nguyện vọng về miền tây': 'Đổi khu vực mong muốn',
}
ops['Nhom_Ly_Do_Rot_PV2'] = ops['KQ_PV_Vong2'].map(reason_map)
ops.loc[ops['Dat_PV2'], 'Nhom_Ly_Do_Rot_PV2'] = np.nan

print(ops['Nhom_Ly_Do_Rot_PV2'].value_counts())
```

```python
unmapped = ops[ops['Co_PV2'] & ~ops['Dat_PV2'] & ops['Nhom_Ly_Do_Rot_PV2'].isna()]['KQ_PV_Vong2'].unique()
print(unmapped)
```

```python
onboard_map = {
    'đã nhận việc': 'Đã nhận việc', 'Đã nhận việc': 'Đã nhận việc',
    'nhận việc': 'Đã nhận việc', 'Nhận việc': 'Đã nhận việc',
    'Có': 'Đã nhận việc', 'CÓ': 'Đã nhận việc',
    'không nhận việc': 'Không nhận việc', 'Không nhận việc': 'Không nhận việc',
    'Không đến nhận việc': 'Không nhận việc', 'Không': 'Không nhận việc',
    'Nghỉ sau 1 ngày OB': 'Nghỉ sớm sau Onboard', 'out 2 ngày': 'Nghỉ sớm sau Onboard',
    'Tai nạn xe gãy chân': 'Sự cố bất khả kháng',
}
ops['Onboard_Note'] = ops['Onboard_Note_raw'].apply(norm_text).map(onboard_map)
ops['Onboard_Note'] = ops['Onboard_Note'].fillna(ops['Onboard_Note_raw'].apply(norm_text))

conditions = [
    ops['Onboard_Note'].isin(['Đã nhận việc']),
    ops['Onboard_Note'].isin(['Không nhận việc']),
    ops['Onboard_Note'].isin(['Nghỉ sớm sau Onboard', 'Sự cố bất khả kháng']),
    ops['Pass_Vong2_Flag'] == 'Pass',
]
choices = ['Có', 'Không', 'Có (nghỉ sớm)', 'Có (suy ra từ Pass PV2)']
ops['Den_Nhan_Viec'] = np.select(conditions, choices, default='Không rõ')

print(ops['Den_Nhan_Viec'].value_counts())
```

##CASE2: TIME-TO-HIRE

```python
ops['TTH_UngTuyen_SoVan'] = (ops['Ngay_So_Van'] - ops['Ngay_Ung_Tuyen']).dt.days
ops['TTH_SoVan_PV2']      = (ops['Ngay_PV_Vong2'] - ops['Ngay_So_Van']).dt.days
ops['TTH_UngTuyen_PV2']   = (ops['Ngay_PV_Vong2'] - ops['Ngay_Ung_Tuyen']).dt.days
ops['TTH_UngTuyen_NhanViec'] = (ops['Ngay_Nhan_Viec'] - ops['Ngay_Ung_Tuyen']).dt.days

for c in ['TTH_UngTuyen_SoVan', 'TTH_SoVan_PV2', 'TTH_UngTuyen_PV2', 'TTH_UngTuyen_NhanViec']:
    n_negative = (ops[c] < 0).sum()
    print(c, "- số dòng âm (lỗi):", n_negative)
    ops.loc[ops[c] < 0, c] = np.nan
```

##CASE3: HIỆU QUẢ KÊNH TUYỂN DỤNG

```python
fb_raw = pd.read_excel(path_file1, sheet_name='FACEBOOK')

COL_HOTEN = 'Họ và tên:'
COL_PHONE = [c for c in fb_raw.columns if c.startswith('Số điện thoại liên hệ')][0]
COL_EMAIL = [c for c in fb_raw.columns if c.startswith('Email:')][0]
COL_DIA_DIEM_MONGMUON = [c for c in fb_raw.columns if c.startswith('Địa điểm mong muốn')][0]
COL_KINHNGHIEM = [c for c in fb_raw.columns if c.startswith('Kinh nghiệm')][0]

fb = fb_raw[fb_raw[COL_HOTEN].notna()].copy()
print(f"FACEBOOK: {len(fb_raw)} dòng thô -> {len(fb)} dòng có dữ liệu thật")
```

```python
fb['SDT'] = fb[COL_PHONE].apply(clean_phone)
fb['Ngay_Nop_Form'] = pd.to_datetime(fb['Dấu thời gian'], errors='coerce')

def clean_source(v):
    if pd.isna(v): return np.nan
    v = str(v).strip()
    if re.fullmatch(r'\d{8,11}', v):
        return 'Không xác định (lỗi nhập liệu)'
    return v

fb['Nguon_Chuan'] = fb['Nguồn'].apply(clean_source)
print("Số dòng nguồn bị lỗi (SĐT nhập nhầm cột):", (fb['Nguon_Chuan']=='Không xác định (lỗi nhập liệu)').sum())
```

```python
fb_sorted = fb.sort_values('Ngay_Nop_Form')
dup_count = fb_sorted['SDT'].duplicated(keep=False).sum()
print(f"Số dòng SĐT bị trùng (nộp form nhiều lần): {dup_count}")

fb_dedup = fb_sorted.drop_duplicates(subset='SDT', keep='last').copy()
print(f"Sau dedup theo SĐT: {len(fb_dedup)} ứng viên duy nhất")
```

```python
sdt_trung = fb_sorted[fb_sorted['SDT'].duplicated(keep=False)]['SDT'].iloc[0]
print(fb_sorted[fb_sorted['SDT']==sdt_trung][['Ngay_Nop_Form','Vị trí ứng tuyển:','Tình trạng']])
```

```python
fb_final_cols = {
    'TA Incharge': 'TA_Incharge', 'Tình trạng': 'Tinh_Trang_XuLy', 'Brand': 'Brand',
    COL_HOTEN: 'Ho_Ten', 'SDT': 'SDT', 'Ngay_Nop_Form': 'Ngay_Nop_Form',
    'Giới tính:': 'Gioi_Tinh', COL_EMAIL: 'Email',
    'Chỗ ở hiện tại': 'Quan_Hien_Tai', 'Vị trí ứng tuyển:': 'Vi_Tri_Ung_Tuyen',
    'Ca làm việc mong muốn:': 'Ca_Mong_Muon', COL_DIA_DIEM_MONGMUON: 'Dia_Diem_Mong_Muon',
    COL_KINHNGHIEM: 'Kinh_Nghiem',
    'Nguon_Chuan': 'Nguon_Chuan', 'Bạn biết thông tin tuyển dụng từ đâu?': 'Nguon_Tu_Khai',
}
fb_clean = fb_dedup[list(fb_final_cols.keys())].rename(columns=fb_final_cols)

merge_case3 = ops.merge(
    fb_clean[['SDT','Ngay_Nop_Form','Nguon_Chuan','Nguon_Tu_Khai','Vi_Tri_Ung_Tuyen']],
    on='SDT', how='left', suffixes=('', '_FB')
)
matched = merge_case3['Ngay_Nop_Form'].notna().sum()
print(f"Khớp được: {matched}/{len(merge_case3)} ({matched/len(merge_case3):.1%})")
```

```python
merge_case3['Nguon_Lech'] = (
    merge_case3['Nguồn ứng viên'].notna() & merge_case3['Nguon_Chuan'].notna() &
    (merge_case3['Nguồn ứng viên'].str.strip() != merge_case3['Nguon_Chuan'].str.strip())
)
print("Số dòng nguồn lệch:", merge_case3['Nguon_Lech'].sum())

print(merge_case3[merge_case3['Nguon_Lech']][['Nguồn ứng viên','Nguon_Chuan']].value_counts().head(10))
```
